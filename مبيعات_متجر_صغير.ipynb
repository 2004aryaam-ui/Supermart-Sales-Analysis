```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Supermart Grocery Sales - Retail Analytics Dataset.csv")

print(df.head())
print(df.info())
```

```python

# تحويل الأعمدة الرقمية إلى أرقام
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce")
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce")
```

```python

# حساب إجمالي المبيعات
total_sales = df["Sales"].sum()
total_sales
```

```python
# الوصف الإحصائي لقيمة المبيعات
sales_description = df["Sales"].describe()
sales_description
```

```python
# حساب متوسط قيمة الطلب
average_sales = df["Sales"].mean()
average_sales
```

```python
# حساب الوسيط
median_sales = df["Sales"].median()
median_sales
```

```python

# حساب أقل وأعلى قيمة
minimum_sales = df["Sales"].min()
maximum_sales = df["Sales"].max()
print("أقل قيمة طلب:", minimum_sales)
print("أعلى قيمة طلب:", maximum_sales)
```

```python
# حساب إجمالي الأرباح
total_profit = df["Profit"].sum()
print("إجمالي الأرباح:", total_profit)
```
